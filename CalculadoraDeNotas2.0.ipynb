```python
nome = input("Digite seu nome: ")
notaMatematica = int(input("Digite sua nota em Matemática: "))
notaPortugues = int(input("Digite sua nota em Português: "))
notaHistoria = int(input("Digite sua nota em História: "))

mediaNotas = (notaMatematica + notaPortugues + notaHistoria) / 3

print("Matematica", notaMatematica)
print("Portugues", notaPortugues)
print("História", notaHistoria)
print(f"Média de notas: {mediaNotas:.2f}")
```
