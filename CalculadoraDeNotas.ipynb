```python
nota1 = 5
nota2 = 8
nota3 = 6

media = (nota1 + nota2 + nota3) / 3

print("Matematica", nota1)
print("Portugues", nota2)
print("História", nota3)
print(f"Média de notas: {media:.2f}")
```
